## Object oriented Programming 

`Whenever you call print() on an object, or use the str() function to convert it into a string, Python automatically looks for the __str__ method inside that object's class to know what to display.``
If you deleted the __str__ method from your class and ran this code:
What happens: Python doesn't naturally know how you want a Student object to look when printed. Because you haven't given it instructions, it falls back to its default behavior.
It prints the module name (__main__), the class name (Student), and the exact location in your computer's memory (0x7f...) where that specific object is stored. While accurate, this is completely useless to a normal user reading your program's output.`

```python
class Student:
    def __init__(self , name , grade):
        self.name = name 
        self.grades = grade
    def average(self):
        return sum(self.grades)/ len(self.grades)

    def __str__(self):
        return f"The name of the person is {self.name} , and his grades are {self.grades}"

    
    def __repr__(self):
        return (f"<person {self.name} , {self.grades}")
```

```python
std = Student("aman" , (80 , 90, 98,70,40))
```

```python
std.average()
```

```python
print(std)
```

```python
print(std.__repr__())
```

```python
class Book:
    Types = ("Hardcover" , "Softcover") ; 
    def __init__(self , name , book_type , weight):
        self.name = name ; 
        self.book_type = book_type; 
        self.weight = weight ; 

    def __repr__(self):
        return f"<Book : {self.name} , Type: {self.book_type} , weighing {self.weight}>"

    @classmethod
    def hard_cover(cls , name , page_weight):
        return cls(name , Book.Types[0] , page_weight + 100)

    @classmethod
    def soft_cover(cls , name , weight):
        return cls(name , Book.Types[1] , weight)
```

```python
book = Book.hard_cover("harrypotter" , 1500)
book_soft = Book.soft_cover("panchtantra" , 200)
```

```python
print(book)
```

```python
print(book_soft)
```

## Class inheritance

```python
class Device:
    def __init__(self , name , connected_by):
        self.name = name ; 
        self.connected_by = connected_by ; 
        self.connected = True
    def __str__(self):
        return f"Device: {self.name!r} , ({self.connected_by})" ; 
    def disconnect(self):
        self.connected = False ; 
        print("Disconnected!!")


class Printer(Device):
    def __init__(self , name , connected_by , capacity):
        super().__init__(name , connected_by)
        self.capacity = capacity ; 
        self.remaining_pages = capacity 

    def __str__(self):
        return f"{super().__str__()} , {self.capacity} , {self.remaining_pages}" ; 

    def print(self , pages):
        if not self.connected:
            print("device not connected!!") ; 
            return ; 
        if (self.remaining_pages- pages < 0):
            print("Printer does not have sufficient pages left !!") ; 
            return ; 
        print(f"Printing {pages} pages") ; 
        self.remaining_pages -= pages ; 
        
```

```python
printer = Printer("Printer" , "usb" , 100)
```

```python
print(printer)
```

```python
printer.print(50)
```

```python
printer.print(70)
```

```python
printer.remaining_pages
```

## Custom Error Classes 

```python
class TooManyPagesError(Exception):
    pass ; 
    
class Book:
    def __init__(self , name: str , page_count:int):
        self.name = name ;
        self.page_count = page_count ;
        self.pages_read = 0 ; 
    def __repr__(self):
        return (
            f"<Book> {self.name} , <Read> {self.pages_read} out of {self.page_count} pages"
        )
    def read(self , pages: int):
        if (self.pages_read + pages > self.page_count):
            
            raise TooManyPagesError(
                f"You have tried to read more number of pages than the total pages in the book which is impossible!!!!"
            )
            
        self.pages_read += pages ;
        print(repr(self)) ; 
       
```

```python
b1 = Book('harrypotter' , 300)
```

```python
b1
```

```python
try:
    b1.read(20) ; 
    b1.read(2000) ; 

except TooManyPagesError as E:
    print(E)
```

```python
def search (sequence , expected , finder):
    for elem in sequence:
        if finder(elem) == expected:
            return elem
    raise RuntimeError(f"Could not find the element {expected}")
        
```

```python
sequence = [
    {"name" : "Aman" , "age": 24} , 
    {"name" : "Pawan" , "age" : 20} , 
    {"name" : "akash" , "age" : 18}
]
```

```python
def get_friend_name(friend):
    return friend["name"]
```

```python
print(search(sequence , "Aman" , get_friend_name))

## another way is to use a lamda function 
print(search(sequence , "Pawan" , lambda friend : friend["name"])) ; 
```

```python
from operator import itemgetter
print(search(sequence , "Pawan" , itemgetter("name")));                  
```
